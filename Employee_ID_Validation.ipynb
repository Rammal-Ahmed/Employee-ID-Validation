```python
# ------------------------
# Employee ID Validation
# ------------------------
# Format: "EMP" followed by exactly 5 digits (e.g. EMP12345)

print("=" * 40)
print("EMPLOYEE ID VALIDATOR".center(40))
print("=" * 40)
print(

)
employee_id = input("Enter Employee ID: ").strip().upper()

if not employee_id.startswith("EMP"):
    print()
    print("Invalid: ID must start with 'EMP'")
elif len(employee_id) != 8:
    print()
    print("Invalid: ID must be exactly 8 characters (EMP + 5 digits)")
elif not employee_id[3:].isdigit():
    print()
    print("Invalid: only digits are allowed after 'EMP'")
else:
    print()
    print(f"Valid: {employee_id} is a correct Employee ID")

print()
print("=" * 40)
print("BUILT FOR THE COMMUNITY HOPE IT HELPS".center(40))
print("SEE YOU IN THE NEXT REPOSITORIES..!".center(40))
print("=" * 40)
```
